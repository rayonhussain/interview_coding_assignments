# Analysis on Netflix Data - Yipit Assignment

## Config and Insertion of the Data

```python
import pandas as pd
import sqlite3
import re

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", None)

file_name = "Data File.xlsx"
netflix = pd.read_excel(file_name, sheet_name="NFLX Top 10")
imdb    = pd.read_excel(file_name, sheet_name="IMDB Rating")

# having title cleanup as it can be interpreted as a different datatype
imdb["title"] = imdb["title"].where(imdb["title"].isna(), imdb["title"].astype(str)) 

OUTAGE_WEEK = pd.Timestamp("2022-05-22")   #this is for outage week as mentioned in the Q/brief

#getting the row/column from the dataframes
print(netflix.shape, imdb.shape) 
netflix.head()

#netflix has 520 rows and 8 columns
#imdb has 15879 rows and 2 columns (predominantly because of higher number of titles/shows overall)
```

## Exploring Data(EDA)

```python
#finding the number of weeks and date range for the netflix data.
print("Weeks:", netflix["week"].dt.date.nunique(),
      "| earliest_date:", netflix["week"].min().date(), 
      "| late_date:", netflix["week"].max().date())

#crosschecking if the categories as defined in the Q's is appropriate or if there are any additional noise
print(netflix["category"].value_counts(), "\n")

# Checking if the outage week is the only affected week
print("Weeks with 0 hours:", netflix.loc[netflix.weekly_hours_viewed == 0, "week"].dt.date.unique())

# IMDb checks on unclean and noise data.
print("\nIMDb null titles     :", imdb["title"].isna().sum())
print("IMDb rating == 0 rows :", (imdb["rating"] == 0).sum(), " <- looks like a 'missing' placeholder")
print("IMDb duplicate titles :", imdb["title"].dropna().duplicated().sum())
```

### Questions and Responses

---
#### Question 1
**As of the most recent week, which English title had the highest cumulative weeks in the top 10? What were its average weekly hours viewed across the whole data set (ignoring the outage week)?**

**Flow is as below:**

*Find the most recent week(max week available) **->** Keep/Filtering English categories only (`Films (English)` and `TV (English)`) **->** Pick the title with the largest `cumulative_weeks_in_top_10` that week (one title can have several season rows, so taking the max per title) **->** For that title, drop the outage week, add up its season rows per week, then average across weeks.*

```python
#max week availble
latest_week = netflix["week"].max()
print("Most recent week:", latest_week.date())

#filtering over english movies and tv series
english_latest = netflix[
    (netflix["week"] == latest_week)
    & (netflix["category"].isin(["Films (English)", "TV (English)"]))
]

#now grouping over the titles and using cumulative weeks column to get max of each title and then sorting in desc order(highest to lowest)
sorted_english_titles = (english_latest.groupby("show_title")["cumulative_weeks_in_top_10"]
       .max().sort_values(ascending=False))
print(sorted_english_titles.head(5))

top_title = sorted_english_titles.index[0] # getting top ranked series/film in english
print("\n Title with most cumulative weeks in top 10:", top_title,": ", sorted_english_titles.iloc[0], "weeks")
```

```python
#now filtering over the top title for all/complete dataset and finding the average by excluding the outage week(impacts average)
t = netflix[(netflix["show_title"] == top_title) & (netflix["week"] != OUTAGE_WEEK)]

#this is to calculate the weekly totals i.e houres watched. As the columns are already present we can aggregate it directly
weekly_total = t.groupby("week")["weekly_hours_viewed"].sum()
print(weekly_total)

#getting average hours watched for top title and its mean/avg.
avg_weekly_hours = weekly_total.mean()
print(f"\nWeeks counted: {len(weekly_total)}")
#just using f string and getting the number of hours watched globally in Millions(hours)
print(f"Average weekly hours viewed for '{top_title}': {avg_weekly_hours:,.0f}  (~{avg_weekly_hours/1e6:,.1f}M)")
```

**Q1 answer:** *Stranger Things* (6 cumulative weeks), averaging about 316.2M hours per week over the 5 non-outage weeks

#### Question 2
**What is the weekly rank of the lowest IMDb-rated title in the most recent week?**

**Flow is as below**
- Take the Netflix rows for the most recent week.
- Join to the IMDb sheet on title (`show_title = title`). **->** Clean the IMDb data first, otherwise the answer is wrong:
   * `rating == 0` is a placeholder for "not rated" this is just hypothesis as the Money Heist: Korea can't be 0.0, 
   * Null titles can't be joined, so dropping them(so join doesnt fail)
   * Duplicate titles (e.g. *The Mist* has 2 ratings) would duplicate Netflix rows, so collapse to one row per title (min rating) (would have been helpful to have date)
- Sort ascending by rating and take the first row. `weekly_rank` is the rank *within its category*.

```python
# --- clean IMDb ---
imdb_clean = (imdb
    .dropna(subset=["title"])            # can't join on a null title
    .query("rating > 0")                 # 0.0 = missing placeholder
    .groupby("title", as_index=False)["rating"].min()   # one row per title
    .rename(columns={"title": "show_title", "rating": "imdb_rating"})) #this is for personal clarity

# filtering over the latest week for joining over the imdb data
latest_week_netflix = netflix[netflix["week"] == latest_week]
merged = latest_week_netflix.merge(imdb_clean, on="show_title", how="left")

print("Latest-week rows:", len(latest_week_netflix),
      "| with a valid IMDb rating:", merged["imdb_rating"].notna().sum())

print("\nNo valid rating:", sorted(merged.loc[merged.imdb_rating.isna(), "show_title"].unique())) 
#showing the titles that might not have rating
```

```python
# --- lowest rated ---
rated = merged.dropna(subset=["imdb_rating"])
lowest = rated[rated["imdb_rating"] == rated["imdb_rating"].min()]   # keeps ties, if any

lowest[["category", "show_title", "imdb_rating", "weekly_rank"]]
```

```python
# Bottom 5 for context, so the answer can be sanity-checked
(rated.sort_values("imdb_rating")
      [["category", "show_title", "imdb_rating", "weekly_rank"]].head(5))
```

**Q2 answer:** the lowest rated title with a valid rating is *Collision* (IMDb 3.8), at **weekly rank 7** in *Films (English)*.
